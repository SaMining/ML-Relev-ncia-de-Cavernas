```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



df_v3 = pd.read_excel(
    "../datasets/resultado_SRA_e_serra_norte_v3.xlsx"
)

df_v4 = pd.read_excel(
    "../datasets/resultado_SRA_e_serra_norte_v4.xlsx"
)

modelos = {
    "V3 - Prospecção": df_v3,
    "V4 - Prospecção + topo": df_v4,
}


thresholds = np.arange(0, 0.51, 0.05)

resultados = []


for nome_modelo, df in modelos.items():


    real_maxima = (
        df["target_relatorio"]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("max")
    )

    quantidade_real_maxima = real_maxima.sum()

    for threshold in thresholds:


        prevista_maxima = df["score"] >= threshold


        acerto = prevista_maxima & real_maxima

        alarme_falso = prevista_maxima & ~real_maxima


        maxima_nao_encontrada = ~prevista_maxima & real_maxima

        resultados.append({
            "modelo": nome_modelo,
            "threshold": threshold,

            "maximas_reais": int(quantidade_real_maxima),

            "previstas_como_maxima": int(
                prevista_maxima.sum()
            ),

            "acertos": int(
                acerto.sum()
            ),

            "alarmes_falsos": int(
                alarme_falso.sum()
            ),

            "maximas_nao_encontradas": int(
                maxima_nao_encontrada.sum()
            ),
        })

metricas = pd.DataFrame(resultados)

# Threshold em percentual, só para exibição
metricas["threshold_percentual"] = (
    metricas["threshold"] * 100
).round().astype(int)

print(metricas)

metricas.to_excel("../datasets/metricas.xlsx", index=False)
```

```python
threshold_escolhido = 0.40

dados_grafico = metricas[
    np.isclose(
        metricas["threshold"],
        threshold_escolhido
    )
].copy()

x = np.arange(len(dados_grafico))
largura = 0.25

fig, ax = plt.subplots(figsize=(11, 6))

barras_reais = ax.bar(
    x - largura,
    dados_grafico["maximas_reais"],
    width=largura,
    label="Máximas reais",
    color="#A5A5A5"
)

barras_previstas = ax.bar(
    x,
    dados_grafico["previstas_como_maxima"],
    width=largura,
    label="Previstas como máxima",
    color="#4472C4"
)

barras_acertos = ax.bar(
    x + largura,
    dados_grafico["acertos"],
    width=largura,
    label="Acertos",
    color="#70AD47"
)
ax.bar_label(barras_reais, padding=3)
ax.bar_label(barras_previstas, padding=3)
ax.bar_label(barras_acertos, padding=3)

ax.set_xticks(x)
ax.set_xticklabels(dados_grafico["modelo"])

ax.set_ylabel("Quantidade de cavidades")

ax.set_title(
    f"Máximas reais, previsões e acertos\n"
    f"Threshold de {threshold_escolhido:.0%}"
)

ax.legend()
ax.grid(axis="y", alpha=0.25)


plt.tight_layout()
plt.savefig(
f"../graficos/previstas_x_acertos/grafico_reais_previstas_acertos_thresh_{round(threshold_escolhido*100)}.png",
dpi=300,
bbox_inches="tight"
)

plt.show()
```

```python
threshold_escolhido = 0.20

dados_grafico = metricas[
    np.isclose(
        metricas["threshold"],
        threshold_escolhido
    )
].copy()

fig, ax = plt.subplots(figsize=(10, 6))

barras_acertos = ax.bar(
    dados_grafico["modelo"],
    dados_grafico["acertos"],
    label="Acertos",
    color="#70AD47"
)

barras_erros = ax.bar(
    dados_grafico["modelo"],
    dados_grafico["alarmes_falsos"],
    bottom=dados_grafico["acertos"],
    label="Alarmes falsos",
    color="#ED7D31"
)

# Rótulos dentro das barras
for indice, linha in dados_grafico.reset_index(drop=True).iterrows():

    # Centro da parte verde
    ax.text(
        indice,
        linha["acertos"] / 2,
        f'{linha["acertos"]} acertos',
        ha="center",
        va="center",
        color="white",
        fontweight="bold"
    )

    # Centro da parte laranja
    if linha["alarmes_falsos"] > 0:
        ax.text(
            indice,
            linha["acertos"]
            + linha["alarmes_falsos"] / 2,
            f'{linha["alarmes_falsos"]} falsos',
            ha="center",
            va="center",
            color="white",
            fontweight="bold"
        )

    # Total acima da barra
    ax.text(
        indice,
        linha["previstas_como_maxima"] + 1,
        f'Total: {linha["previstas_como_maxima"]}',
        ha="center",
        va="bottom",
        fontweight="bold"
    )

ax.set_ylabel("Quantidade prevista como máxima")

ax.set_title(
    f"Composição das previsões de máxima\n"
    f"Threshold de {threshold_escolhido:.0%}"
)

ax.legend()
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.savefig(
f"../graficos/composicao_previsoes/composicao_prev_max_thresh_{round(threshold_escolhido*100)}.png",
dpi=300,
bbox_inches="tight"
)
plt.show()
```

```python
threshold_escolhido = 0.20

dados_grafico = metricas[
    np.isclose(
        metricas["threshold"],
        threshold_escolhido
    )
].copy()

fig, ax = plt.subplots(figsize=(10, 6))

barras_encontradas = ax.bar(
    dados_grafico["modelo"],
    dados_grafico["acertos"],
    label="Máximas encontradas",
    color="#4472C4"
)

barras_nao_encontradas = ax.bar(
    dados_grafico["modelo"],
    dados_grafico["maximas_nao_encontradas"],
    bottom=dados_grafico["acertos"],
    label="Máximas não encontradas",
    color="#C9C9C9"
)

for indice, linha in dados_grafico.reset_index(drop=True).iterrows():

    percentual_encontrado = (
        linha["acertos"]
        / linha["maximas_reais"]
        * 100
    )

    percentual_nao_encontrado = (
        linha["maximas_nao_encontradas"]
        / linha["maximas_reais"]
        * 100
    )

    # Rótulo das encontradas
    ax.text(
        indice,
        linha["acertos"] / 2,
        (
            f'{linha["acertos"]}\n'
            f'({percentual_encontrado:.1f}%)'
        ),
        ha="center",
        va="center",
        color="white",
        fontweight="bold"
    )

    # Rótulo das não encontradas
    ax.text(
        indice,
        linha["acertos"]
        + linha["maximas_nao_encontradas"] / 2,
        (
            f'{linha["maximas_nao_encontradas"]}\n'
            f'({percentual_nao_encontrado:.1f}%)'
        ),
        ha="center",
        va="center",
        color="black",
        fontweight="bold"
    )

ax.set_ylabel("Quantidade de máximas reais")

ax.set_title(
    f"Cobertura das cavidades máximas\n"
    f"Threshold de {threshold_escolhido:.0%}"
)

ax.legend()
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig(
f"../graficos/cobertura/cobertura_thresh_{round(threshold_escolhido*100)}.png",
dpi=300,
bbox_inches="tight"
)
plt.show()
```

```python
nomes_modelos = metricas["modelo"].unique()

fig, eixos = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(15, 6),
    sharey=True
)

for eixo, nome_modelo in zip(eixos, nomes_modelos):

    dados_modelo = (
        metricas[
            metricas["modelo"] == nome_modelo
        ]
        .sort_values("threshold")
    )

    # Linha com a quantidade prevista como máxima
    eixo.plot(
        dados_modelo["threshold_percentual"],
        dados_modelo["previstas_como_maxima"],
        marker="o",
        linewidth=2.5,
        color="#ED7D31",
        label="Previstas como máxima"
    )

    # Linha com os acertos
    eixo.plot(
        dados_modelo["threshold_percentual"],
        dados_modelo["acertos"],
        marker="o",
        linewidth=2.5,
        color="#70AD47",
        label="Máximas encontradas"
    )

    # Linha horizontal com o total real de máximas
    quantidade_real = dados_modelo[
        "maximas_reais"
    ].iloc[0]

    eixo.axhline(
        y=quantidade_real,
        color="#7F7F7F",
        linestyle="--",
        linewidth=2,
        label=f"Máximas reais: {quantidade_real}"
    )

    eixo.set_title(nome_modelo)
    eixo.set_xlabel("Threshold (%)")
    eixo.set_xticks(range(0, 51, 5))

    eixo.grid(alpha=0.25)
    eixo.legend()

eixos[0].set_ylabel("Quantidade de cavidades")

fig.suptitle(
    "Efeito do threshold sobre previsões e acertos",
    fontsize=16,
    fontweight="bold"
)
plt.tight_layout()
plt.savefig(
f"../graficos/quantidades_por_threshold/quantidades_por_threshold.png",
dpi=300,
bbox_inches="tight"
)
plt.tight_layout()
plt.show()
```
